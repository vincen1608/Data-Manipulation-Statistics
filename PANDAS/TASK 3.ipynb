```python
#filtering is process of selecting rows that satisfy specific conditions 
#filter products with unit price greater price
high_price = df[df['Unit_Price'] > 1000]

high_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
#filtering is process of selecting rows that satisfy specific conditions 
#filter products with unit price lesser price
low_price = df[df['Unit_Price'] < 500]

low_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
#Filter Products with price between 500 and 1500

medium_price = df[
    (df['Unit_Price'] >= 500) &
    (df['Unit_Price'] <= 1500)
]

medium_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
#Filter Products with price between 500 and 1500

medium_price = df[
    (df['Unit_Price'] >= 500) &
    (df['Unit_Price'] <= 1500)
]

medium_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
#ascending function
sorted_price = df.sort_values(
    by='Unit_Price',
)
sorted_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
sorted_price = df.sort_values(
    by='Unit_Price',
    ascending=False #decending function
)
sorted_price[['Product_Name', 'Category', 'Unit_Price']].head()
```
